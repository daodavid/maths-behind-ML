# Continuous Distributions

## Uniform Distribution

The Uniform Distribution is a continuous probability distribution in which all values within a given interval have an equal probability density.

In simple words, the distribution is uniform because values are spread evenly across an interval.

It is defined by two parameters:
- \(a\): Lower limit (minimum value).
- \(b\): Upper limit (maximum value).
We write:
\[
X\sim U(a,b)
\]

Where \(U\) represents the Uniform Distribution.

Probability Density Function (PDF)
The PDF of a continuous Uniform Distribution is:
\[
f(x)=
\begin{cases}
\frac{1}{b-a}, & a\leq x\leq b\\
0, & \text{otherwise}
\end{cases}
\]

## Normal Distribution

3.1 Definition
The Normal Distribution is a continuous probability distribution that is symmetrical and bell-shaped, where most observations are concentrated around the mean.

It is also known as the Gaussian Distribution.

A normal random variable is written as:
\[
X\sim N(\mu,\sigma^2)
\]

Where:
- \(X\) = Random variable.
- \(\mu\) = Mean of the distribution.
- \(\sigma\) = Standard deviation.
- \(\sigma^2\) = Variance.

The normal distribution has the following characteristics:
1. It is bell-shaped.
2. It is symmetrical around the mean.
3. The mean, median and mode are equal.
4. The total area under the curve is equal to 1.
5. The curve extends indefinitely in both directions.
6. Most observations are concentrated around the mean.

3.3 Probability Density Function (PDF)
The PDF of a Normal Distribution is given by:
\[
\boxed{
f(x)=\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
}
\]

Where:
- \(e\) = Mathematical constant approximately 2.71828.
- \(\pi\) = Mathematical constant approximately 3.14159.
- \(\mu\) = Mean.
- \(\sigma\) = Standard deviation.
Note: The normal PDF is defined for all real numbers, from \(-\infty\) to \(+\infty\).
3.4 Standard Normal Distribution
A Standard Normal Distribution is a special type of normal distribution with:
\[
\boxed{\mu=0,\quad \sigma=1}
\]

It is written as:
\[
Z\sim N(0,1)
\]

To convert an ordinary normal variable into a standard normal variable, we use the Z-score formula:
\[
\boxed{Z=\frac{X-\mu}{\sigma}}
\]

The Z-score tells us how many standard deviations a value is above or below the mean.
